```python
import pandas as pd
import numpy as np

df = pd.read_csv("exported_churn_data.csv")

print("Dataset loaded successfully!")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
```

```python
# Display first 5 rows

df.head()
```

```python
# Check dataset structure

df.info()
```

```python
# Check number of rows and columns

print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])
```

```python
# Check missing values

missing_values = df.isnull().sum()

print("Missing values in each column:")
print(missing_values)
```

```python
# Calculate missing percentage

missing_percentage = (df.isnull().sum() / len(df)) * 100

missing_summary = pd.DataFrame({
    "Missing Values": df.isnull().sum(),
    "Missing Percentage": missing_percentage
})

missing_summary
```

```python
# Check duplicate records

duplicate_count = df.duplicated().sum()

print("Number of duplicate records:", duplicate_count)
```

```python
# Check current data types

df.dtypes
```

```python
date_columns = [
    "subscription_start_date",
    "renewal_date",
    "cancellation_date",
    "dob",
    "complaint_date"
]

for column in date_columns:
    df[column] = pd.to_datetime(df[column], errors="coerce")

print("Date columns converted successfully.")
```

```python
df[date_columns].dtypes
```

```python
df.columns = (
    df.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
)

df.columns
```

```python
# Customers without cancellation have no cancellation reason/date
df["cancellation_reason"] = df["cancellation_reason"].fillna("Not Cancelled")

# Complaint-related fields
df["complaint_count"] = df["complaint_count"].fillna(0)
df["escalations"] = df["escalations"].fillna(0)
df["csat_score"] = df["csat_score"].fillna(0)
```

```python
print("Missing values after cleaning:")

df.isnull().sum()
```

```python
df.to_csv("cleaned_customer_data.csv", index=False)

print("Cleaned dataset saved successfully!")
```
