```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load the dataset
df = pd.read_csv("exported_churn_data.csv")

# Display dataset information
print("Dataset loaded successfully!")
print("Dataset Shape:", df.shape)

print("\nColumn Names:")
print(df.columns.tolist())

# Display first five rows
df.head()

```

```python

# Customer Churn Visualization

plt.figure(figsize=(8, 5))

sns.countplot(
    data=df,
    x="Churn Flag",
    hue="Churn Flag",
    palette="Set2",
    legend=False
)

plt.title("Customer Churn Analysis", fontsize=16, fontweight="bold")
plt.xlabel("Churn Status", fontsize=12)
plt.ylabel("Number of Customers", fontsize=12)

plt.tight_layout()
plt.show()
```

```python

# Contract Type vs Customer Churn

plt.figure(figsize=(9, 6))

sns.countplot(
    data=df,
    x="contract_type",
    hue="Churn Flag",
    palette="Set1"
)

plt.title("Contract Type vs Customer Churn", fontsize=16, fontweight="bold")
plt.xlabel("Contract Type", fontsize=12)
plt.ylabel("Number of Customers", fontsize=12)
plt.xticks(rotation=30)
plt.legend(title="Churn Status")

plt.tight_layout()
plt.show()
```

```python

# Monthly Charges Visualization

plt.figure(figsize=(9, 6))

sns.histplot(
    data=df,
    x="monthly_charges",
    bins=25,
    kde=True,
    color="steelblue"
)

plt.title("Distribution of Monthly Charges", fontsize=16, fontweight="bold")
plt.xlabel("Monthly Charges", fontsize=12)
plt.ylabel("Number of Customers", fontsize=12)

plt.tight_layout()
plt.show()
```

```python

# Customer Cancellation Reasons Visualization

plt.figure(figsize=(10, 6))

reason_counts = df["cancellation_reason"].dropna().value_counts()

sns.barplot(
    x=reason_counts.values,
    y=reason_counts.index,
    color="cornflowerblue"
)

plt.title("Most Common Customer Cancellation Reasons",
          fontsize=16, fontweight="bold")
plt.xlabel("Number of Customers", fontsize=12)
plt.ylabel("Cancellation Reason", fontsize=12)

plt.tight_layout()
plt.show()
```

```python

# Churn Score Visualization

plt.figure(figsize=(9, 6))

sns.histplot(
    data=df,
    x="churn_score",
    bins=25,
    kde=True,
    color="mediumpurple"
)

plt.title("Customer Churn Score Distribution",
          fontsize=16, fontweight="bold")
plt.xlabel("Churn Score", fontsize=12)
plt.ylabel("Number of Customers", fontsize=12)

plt.tight_layout()
plt.show()
```

```python

import os
import matplotlib.pyplot as plt
import seaborn as sns

# Create folder
os.makedirs("graphs", exist_ok=True)

# Graph 1: Customer Churn
plt.figure(figsize=(8, 5))
sns.countplot(data=df, x="Churn Flag", hue="Churn Flag",
              palette="Set2", legend=False)
plt.title("Customer Churn Analysis")
plt.xlabel("Churn Status")
plt.ylabel("Number of Customers")
plt.tight_layout()
plt.savefig("graphs/customer_churn.png", dpi=300, bbox_inches="tight")
plt.close()

# Graph 2: Contract Type vs Churn
plt.figure(figsize=(9, 6))
sns.countplot(data=df, x="contract_type", hue="Churn Flag",
              palette="Set1")
plt.title("Contract Type vs Customer Churn")
plt.xlabel("Contract Type")
plt.ylabel("Number of Customers")
plt.xticks(rotation=30)
plt.tight_layout()
plt.savefig("graphs/contract_type_vs_churn.png", dpi=300, bbox_inches="tight")
plt.close()

# Graph 3: Monthly Charges
plt.figure(figsize=(9, 6))
sns.histplot(data=df, x="monthly_charges", bins=25, kde=True,
             color="steelblue")
plt.title("Distribution of Monthly Charges")
plt.xlabel("Monthly Charges")
plt.ylabel("Number of Customers")
plt.tight_layout()
plt.savefig("graphs/monthly_charges.png", dpi=300, bbox_inches="tight")
plt.close()

# Graph 4: Cancellation Reasons
reason_counts = df["cancellation_reason"].dropna().value_counts()
plt.figure(figsize=(10, 6))
sns.barplot(x=reason_counts.values, y=reason_counts.index,
            color="cornflowerblue")
plt.title("Customer Cancellation Reasons")
plt.xlabel("Number of Customers")
plt.ylabel("Cancellation Reason")
plt.tight_layout()
plt.savefig("graphs/cancellation_reasons.png", dpi=300, bbox_inches="tight")
plt.close()

# Graph 5: Churn Score
plt.figure(figsize=(9, 6))
sns.histplot(data=df, x="churn_score", bins=25, kde=True,
             color="mediumpurple")
plt.title("Customer Churn Score Distribution")
plt.xlabel("Churn Score")
plt.ylabel("Number of Customers")
plt.tight_layout()
plt.savefig("graphs/churn_score.png", dpi=300, bbox_inches="tight")
plt.close()

print("All 5 graphs saved successfully!")
print("Location: Task-3-Data-Visualization/graphs/")
```

```python

# Pie Chart: Customer Churn Distribution

churn_counts = df["Churn Flag"].value_counts()

plt.figure(figsize=(7, 7))

plt.pie(
    churn_counts.values,
    labels=churn_counts.index,
    autopct="%1.1f%%",
    startangle=90,
    colors=sns.color_palette("Set2", len(churn_counts))
)

plt.title("Customer Churn Distribution", fontsize=16, fontweight="bold")
plt.axis("equal")
plt.tight_layout()
plt.show()
```

```python

# Line Chart: Monthly Charges Trend

monthly_trend = (
    df["monthly_charges"]
    .dropna()
    .sort_values()
    .reset_index(drop=True)
)

plt.figure(figsize=(9, 5))

plt.plot(
    monthly_trend.index,
    monthly_trend.values,
    color="teal",
    linewidth=2
)

plt.title("Monthly Charges Distribution Trend",
          fontsize=16, fontweight="bold")
plt.xlabel("Customer Index (sorted by charges)")
plt.ylabel("Monthly Charges")

plt.tight_layout()
plt.show()
```

```python

# Scatter Plot: Monthly Charges vs Churn Score

plt.figure(figsize=(9, 6))

sns.scatterplot(
    data=df,
    x="monthly_charges",
    y="churn_score",
    hue="Churn Flag",
    alpha=0.7,
    s=60
)

plt.title(
    "Monthly Charges vs Churn Score",
    fontsize=16,
    fontweight="bold"
)
plt.xlabel("Monthly Charges", fontsize=12)
plt.ylabel("Churn Score", fontsize=12)
plt.legend(title="Churn Status")

plt.tight_layout()
plt.show()
```
